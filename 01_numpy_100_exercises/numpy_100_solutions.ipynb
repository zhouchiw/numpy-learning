#### 17. What is the result of the following expression? (★☆☆)

#### 63. Create an array class that has a name attribute (★★☆)

````python
class NamedArray(np.ndarray):
    def __new__(cls, array, name="no name"): # NumPy arrays are created in __new__, not __init__
        obj = np.asarray(array).view(cls)    # Interpret this array as class NamedArray
        obj.name = name
        return obj
    def __array_finalize__(self, obj):
        if obj is None: 
            return
        self.name = getattr(obj, 'name', "no name")

Z = NamedArray(np.arange(10), "range_10")
print (Z.name)
````

#### 75. How to compute averages using a sliding window over an array? (★★★)

````python
# Method 1: cumulative sum (efficient)
def moving_average(arr, window_size=3):
    cumsum = np.cumsum(arr)
    cumsum[window_size:] = cumsum[window_size:] - cumsum[:-window_size]
    return cumsum[window_size - 1:] / window_size
arr = np.arange(10)
averages = moving_average(arr,4)
print(averages)
````

````python
# Method 2: sliding_window_view
from numpy.lib.stride_tricks import sliding_window_view
arr = np.arange(10)
print(sliding_window_view(arr, window_shape=3).mean(axis=-1))
````

````python
# Method 3: convolution
arr = np.arange(10)
n = 3
kernel = np.ones(n)/n
result = np.convolve(arr, kernel, mode='valid')
print(result)
````

#### 80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)

````python
def extract(arr, shape, position, fill=0):
    arr = np.asarray(arr)
    pad_width = [(s//2, s//2) for s in shape]  #[(2,2),(2,2)]
    padded_arr = np.pad(arr, pad_width, mode='constant', constant_values=fill) 
    padded_position = np.array(position) + np.array(shape)//2
    slices_list = []
    for p, s in zip(padded_position, shape):
        start = p - s//2
        stop = start + s
        slices_list.append(slice(start, stop))
    slices = tuple(slices_list)
    return padded_arr[slices]

arr = np.arange(100).reshape(10,10)
result = extract(arr,shape=(5,5),position=(1,1),fill=0)
print(result)
````

````python
class Symmetric(np.ndarray):
    def __new__(cls, input_array):
        obj = np.asarray(input_array).view(cls)
        return obj
    def __setitem__(self, index, value):
        super().__setitem__(index, value)
        if isinstance(index, tuple) and len(index) == 2:
            i, j = index
            super().__setitem__((j, i), value)

S = Symmetric(np.random.randint(0,10,(5,5)))
S[2,3] = 42
print(S)
````

#### 88. How to implement the Game of Life using numpy arrays? (★★★)

````python
# Method 1
from numpy.lib.stride_tricks import sliding_window_view
Z = np.random.randint(0,2,(10,10))
kernel = np.array([[1,1,1],
                   [1,0,1],
                   [1,1,1]])
def gameoflife(Z):
    Z_pad = np.pad(Z, 1, mode = 'wrap')
    W = sliding_window_view(Z_pad, (3,3))
    N = np.sum(W * kernel, axis=(-2, -1))
    birth = (Z == 0) & (N == 3)
    survive = (Z == 1) & ((N == 2) | (N == 3))
    return (birth | survive).astype(Z.dtype)

result = gameoflife(Z)
print(result)
````

````python
# Method 2: Roll function (assume no edge effect)
Z = np.random.randint(0,2,(10,10))

def gameoflife_roll(Z):
    N = (
            np.roll(np.roll(Z, -1, 0), -1, 1) +
            np.roll(Z, -1, 0) +
            np.roll(np.roll(Z, -1, 0), 1, 1) +
            np.roll(Z, 1, 1) +
            np.roll(Z, -1, 1) +
            np.roll(np.roll(Z, 1, 0), -1, 1) +
            np.roll(Z, 1, 0) +
            np.roll(np.roll(Z, 1, 0), 1, 1)
        )
    birth = (Z == 0) & (N == 3)
    survive = (Z == 1) & ((N == 2) | (N == 3))
    return (birth | survive).astype(Z.dtype)

result = gameoflife_roll(Z)
print(result)
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?

````python
import matplotlib.pyplot as plt

phi = np.arange(0, 10*np.pi, 0.1)
a = 1
x = a*phi*np.cos(phi)
y = a*phi*np.sin(phi)
dr = (np.diff(x)**2 + np.diff(y)**2)**.5  # Compute segment lengths
r = np.zeros_like(x)
r[1:] = np.cumsum(dr)                     # Compute cumulative distance
r_int = np.linspace(0, r.max(), 200)      # Create equally spaced value
x_int = np.interp(r_int, r, x)
y_int = np.interp(r_int, r, y)

plt.figure(figsize=(8,8))
plt.plot(x, y, 'grey', label="original curve")
plt.plot(x, y, 'g.', label="original points")
plt.plot(x_int, y_int, 'r.', label="equidistant points")
plt.legend()
plt.axis("equal")
plt.show()
````
